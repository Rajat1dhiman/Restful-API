1. What is a RESTful API?

Ans. A RESTful API(Representational State Transfer) is an architectural style for designing networked appilcations, primarily used in web services. It allows different software systems to communicate with each other over the internet using HTTP . RESTful API follow a set of principles that make them simple, scalable and easy to use.


**Key Characteristics :**

1. **Stateless:** Each request from a client to the server must contain all the information needed to understand and process the request.

2. **Client-Server Architecture:** The client and the server are separate entities that communicate over a network.

3. **Uniform Interface:** RESTful API have a consistent and predictable structure.

4. **Resource-Based:** Everything in REST is considered a resource. A resource could be any piece of data or service that the server provides, and it is accessed via URLs

5. **Representation:** Resources can have different representations. For example, a resource may be represented as JSON, XML or other formats.

2. Explain the concept of API specification.

Ans. An API specification is a document or set of rules that defines how an API behaves, how it should be used, and what developers can expect when interacting with it. It outlines the structure, endpoints , methods, parameters, responses and oter relevant details to ensure both the API provider and consumers understand how to communicate effectively.

**Key Components**

1. **Endpoints:**
        
        * These are the specific URLs that clients use to interact with the API. Each endpoint represents a particular resource or functionality.
        
2. **HTTP Methods:**

        * The specification defines which HTTP methods(such as GET,POST,PUT,DELETE) are allowed for each endpoint.

3. **Request Parameters:**

        * Parameters can be passed in the URL , request body or headers.
        * The specification lists which parameters are required , optional and their data types

4. **Request Body:**
        
        * When making requests that involve sending data, the specification describes the format of the request body.

4. What is routing in Flask?

Ans. In Flask, routing refers to the process of mapping URLs to specific functions in your application. thses functions, known as **view functions**, are executed when a request to a particular URL, and the result is returned to the client.

Flask uses a decorator called **@app.route()** to associate a URL pattern with a view function. When a user accesses a URL that matches a route, the corresponding view function is invoked, and the response is sent back.

```python
#basic routing example

from flask import Flask

app= Flask(__name__)

@app.route('/')
def home():
  return "Hello,World!"

if __name__=='__main__':
  app.run(debug=True)
```

7. What is the purpose of the @app.route() decorator in Flask?

Ans. The **app.route()** decorator in Flask is used to define **routes** for your web application. A **route** is a URL pattern that Flask listens for and when a request matches that pattern, the corresponding function is executed.

In Simpler terms, the @app.route() decorator maps a URL to a specific function in your application. When a user visits that URL, Flask calls the associated functions and sends the returned result back to the client.

**Purpose of @app.route()**:

* **URL Mapping**: The decorator maps a specific URL to a function , which allows Flask to know what should happen when a user vists that URL.

* **Request handling**: It listens for specific HTTP request on a URL and the associated function defines how to respond to that request.

8. What is the difference between GET and POST HTTP methods?

Ans. 1. **Purpose**
* **GET:**
       * The GET method is used to retrieve data from  a server. When you use GET, you are asking the server to send back some information.

       * It is typically used to request data or display a web page.

* **POST:**
        * The POST method is used to send data  to the server, often a **create or update**  a resource.

2. **Data Transmission**
* **GET:**
       * **Data is sent in the URL**. It appends the data to the URL, making it visiblke in the browser's address bar.

* **POST:**
        * **Data is sent in the body ** of the request , not in the URL . THis means  data is not visible in the URL and there is no practical limit in the amount of data that can be sent.

3. **Idempotence:**
* **GET:**
       * **GET request are idompotent,** meaning calling the same GET request multiple times will produce the same result and have no side effects on the server. It is safe for repeated use.

* **POST:**
        * **POST requests are not idempotent**, meaning calling the same POST request multiple times could result in different outcomes

4. **Caching**:
* **GET**
       * GET requests can be cached by browsers or intermediate caches. This can improve performance by reusing the same data for repeated requests.

* **POST:**
        * **POST requests are  not cached** by default, as they are typicallt used for actions that modify data and caching ncould lead to issues with repeated submissions.

7. How can you validate form data in Flask?

```python
#Manual Validation Using request.form
from flask import Flask, render_template, request

app = Flask(__name__)

@app.route('/login', methods=['GET', 'POST'])
def login():
    error = None
    if request.method == 'POST':
        username = request.form.get('username')
        password = request.form.get('password')

        if not username or not password:
            error = "Both fields are required!"
        else:
            return f"Welcome, {username}!"

    return render_template('login.html', error=error)

if __name__ == '__main__':
    app.run(debug=True)
```

8. How do you manage sessions in Flask?

```python
#Enabling Sessions in Flask
from flask import Flask, session, redirect, url_for, request

app = Flask(__name__)
app.config['SECRET_KEY'] = 'supersecretkey'  # Required for session security

@app.route('/')
def home():
    return f"User: {session.get('username', 'Guest')}"

@app.route('/login', methods=['POST'])
def login():
    session['username'] = request.form['username']
    return redirect(url_for('home'))

@app.route('/logout')
def logout():
    session.pop('username', None)  # Remove session data
    return redirect(url_for('home'))

if __name__ == '__main__':
    app.run(debug=True)
```

```python
#Initializing the Flask App (app/__init__.py)
from flask import Flask

def create_app():
    app = Flask(__name__)

    # Register Blueprints
    from app.blueprints.auth import auth_bp
    from app.blueprints.user import user_bp
    app.register_blueprint(auth_bp, url_prefix="/auth")
    app.register_blueprint(user_bp, url_prefix="/user")

    return app
```

12. How do you define a custon Jinja filter in Flask?

```python
#Defining a Custom Filter in Flask
from flask import Flask, render_template

app = Flask(__name__)

# Custom Jinja filter to reverse a string
def reverse_string(s):
    return s[::-1]

# Register the filter with Jinja
app.jinja_env.filters['reverse'] = reverse_string

@app.route('/')
def home():
    return render_template('index.html', name="Flask")

if __name__ == '__main__':
    app.run(debug=True)
```

```python
#Using the Custom Filter in a Jinja Template (templates/index.html)
<!DOCTYPE html>
<html>
<head>
    <title>Custom Jinja Filter</title>
</head>
<body>
    <h1>Original: {{ name }}</h1>
    <h1>Reversed: {{ name | reverse }}</h1>
</body>
</html>
```

```python
#Alternative: Using app.template_filter Decorator
#You can also register a filter using the @app.template_filter() decorator.
@app.template_filter('capitalize_first')
def capitalize_first_letter(s):
    return s.capitalize()
```

```python
#Usage in Jinja Template:
<p>{{ "hello world" | capitalize_first }}</p>
```

13. How do you redirect with query parameters in Flask?

```python
#Basic Redirect with Query Parameters
from flask import Flask, redirect, url_for, request

app = Flask(__name__)

@app.route('/search')
def search():
    query = request.args.get('q', 'No Query Provided')
    return f"Search results for: {query}"

@app.route('/redirect_search')
def redirect_search():
    return redirect(url_for('search', q='Flask Tutorial'))

if __name__ == '__main__':
    app.run(debug=True)
# When visiting /redirect_search, it redirects to /search?q=Flask+Tutorial.
# The request.args.get('q') retrieves the query parameter value.
```

```python
#Redirect with Multiple Query Parameters
@app.route('/results')
def results():
    term = request.args.get('term', 'Default')
    category = request.args.get('category', 'All')
    return f"Showing results for '{term}' in category '{category}'"

@app.route('/redirect_results')
def redirect_results():
    return redirect(url_for('results', term='Python', category='Books'))
```

14. How do you return JSON responses in Flask?

```python
#Using jsonify() (Recommended)
from flask import Flask, jsonify

app = Flask(__name__)

@app.route('/data')
def get_data():
    return jsonify({"message": "Hello, Flask!", "status": "success"})

if __name__ == '__main__':
    app.run(debug=True)
```

```python
#Returning JSON Directly
@app.route('/direct_json')
def direct_json():
    return {"message": "Hello, Flask!", "status": "success"}
#Flask (≥1.1) automatically converts dictionaries to JSON responses.
```

```python
#Customizing Response with Status Code
@app.route('/custom_status')
def custom_status():
    return jsonify({"error": "Unauthorized"}), 401  # HTTP 401 Unauthorized
#This sets a custom HTTP status code.
```

```python
#Using Response() for More Control
from flask import Response
import json

@app.route('/custom_response')
def custom_response():
    data = {"message": "Custom Response"}
    response = Response(json.dumps(data), status=200, mimetype='application/json')
    return response
```
